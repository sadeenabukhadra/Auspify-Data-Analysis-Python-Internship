# Task 5 (Advanced) – Content Rating & Genre Analysis
**Goal:** Examine Netflix content ratings and genres to understand audience targeting.

**Workflow:** Analyze ratings → Identify top genres → Compare ratings across content types → Build interactive visualizations → Present audience insights.

```python
import pandas as pd
import plotly.express as px

import os
path = "/content/Dataset.csv" if os.path.exists("/content/Dataset.csv") else "Dataset.csv"
df = pd.read_csv(path)
```

```python
"""
Step 1: Analyze rating categories.

I used value_counts() to calculate the number of titles in each rating
category. This helps identify the most common content ratings and provides
an initial understanding of the audience categories represented in the
Netflix dataset.
"""
```

## Step 1: Analyze Rating Categories
Count titles in each rating category to understand the audience segments Netflix content targets.

```python
rating_counts = df["rating"].value_counts()
print(rating_counts)
#TV-MA is the most common rating, indicating that a large portion of the content in the dataset is targeted toward mature audiences.
```

```python
"""
Step 2: Identify the most common genres.

The listed_in column can contain multiple genres for a single title.
I split these genre combinations into individual categories using split()
and explode(), then counted their frequency using value_counts().

This allows me to identify the most common genres in the Netflix dataset.
"""
```

## Step 2: Identify the Most Common Genres
The `listed_in` column holds multiple genres per title, so we split and explode it into individual genre rows before counting frequency.

```python
genres = df["listed_in"].str.split(", ").explode()
print(genres)
```

```python
genre_counts = genres.value_counts()
top_genres = genre_counts.head(10).reset_index()
print(top_genres)
```

```python
"""
Step 3: Compare ratings across content types.

I used a cross-tabulation to compare rating categories between Movies and
TV Shows. This allows me to examine how the distribution of content ratings
differs between the two content types.
"""
```

## Step 3: Compare Ratings Across Content Types
Cross-tabulate rating vs. type to see how the rating mix differs between Movies and TV Shows.

```python
rating_by_type = pd.crosstab(df["rating"], df["type"])
print(rating_by_type)
```

```python
# Step 4 — Interactive visualizations
```

## Step 4: Interactive Visualizations
Build interactive Plotly charts for rating distribution, top genres, and ratings by content type.

```python
rating_df = rating_counts.reset_index()

rating_df.columns = ["rating", "content_count"]

fig = px.bar(
    rating_df,
    x="rating",
    y="content_count",
    title="Netflix Content Rating Distribution",
    labels={
        "rating": "Rating",
        "content_count": "Number of Titles"
    }
)

fig.show()
```

```python
# Top Genres
top_genres = genre_counts.head(10).reset_index()

top_genres.columns = ["genre", "content_count"]

fig = px.bar(
    top_genres,
    x="content_count",
    y="genre",
    orientation="h",
    title="Top 10 Netflix Genres",
    labels={
        "genre": "Genre",
        "content_count": "Number of Titles"
    }
)

fig.show()
```

```python
# Ratings by Content Type

rating_type_df = rating_by_type.reset_index()
print(rating_type_df)

fig = px.bar(
    rating_type_df,
    x="rating",
    y=["Movie", "TV Show"],
    barmode="group",
    title="Netflix Rating Distribution by Content Type",
    labels={
        "rating": "Rating",
        "value": "Number of Titles",
        "variable": "Content Type"
         }
)
fig.show()
```

```python
rating_percentage = (
    df["rating"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print(rating_percentage)
#TV-MA represents 36.46% of the content.
```

### Rating & Genre Share (%)
Express ratings and genres as percentages of the total catalog for easier comparison.

```python
genre_counts.head(10)
```

```python
genre_percentage = (
    genre_counts
    .div(len(df))
    .mul(100)
    .round(2)
)

print(genre_percentage.head(10))
```

```python
rating_by_type_percentage = pd.crosstab(
    df["rating"],
    df["type"],
    normalize="columns"
).mul(100).round(2)

print(rating_by_type_percentage)
#TV-MA represents  33.66% of TV Shows compared with  42.91% of Movies.
```

## Audience Insights — Executive Summary

### Rating Analysis
**TV-MA is the single most common rating**, covering **36.46%** of all titles (3,205 of 8,790), followed by **TV-14 (24.54%, 2,157 titles)**. Together these two mature-audience ratings account for **over 60% of the entire catalog**. Family/general-audience ratings — TV-G (2.50%), G (0.47%), TV-Y (3.48%) — make up only a small fraction, confirming that adult and teen viewers are Netflix's primary audience segments rather than young children.

### Genre Analysis
The top genres are **International Movies (2,752 titles, 31.3%)**, **Dramas (2,426, 27.6%)**, **Comedies (1,674, 19.0%)**, **International TV Shows (1,349, 15.4%)**, and **Documentaries (869, 9.9%)**. The strong showing of "International" categories at #1 for both Movies and TV Shows reflects a deliberate global-content sourcing strategy, while Dramas and Comedies anchor the platform's mainstream appeal.

### Content Type and Rating Comparison
The rating mix differs meaningfully by content type: **TV-MA represents 42.91% of TV Shows but only 33.66% of Movies**. In other words, TV Shows skew more heavily toward mature content than Movies do — likely because long-running series (crime dramas, thrillers) more often carry mature ratings, while the movie catalog includes a wider mix of family and general-audience titles alongside its mature-rated films.

### Overall Audience Insight
Netflix's content strategy leans toward **mature, internationally-sourced drama and comedy content**, with TV Shows skewing more mature than Movies. This points to a core audience of teens and adults, a strong emphasis on international storytelling, and a supporting (but smaller) family-content offering. For content strategy, this suggests continued investment in mature dramas/comedies and international titles will resonate with the platform's core viewer base, while family-rated content remains a comparatively underweighted category worth monitoring for growth opportunities.
