```python
import pandas as pd
import plotly.express as px

df = pd.read_csv("/content/Dataset.csv")
```

```python
"""
Step 1: Analyze rating categories.

I used value_counts() to calculate the number of titles in each rating
category. This helps identify the most common content ratings and provides
an initial understanding of the audience categories represented in the
Netflix dataset.
"""
```

```python
rating_counts = df["rating"].value_counts()
print(rating_counts)
#TV-MA is the most common rating, indicating that a large portion of the content in the dataset is targeted toward mature audiences.
```

```python
"""
Step 2: Identify the most common genres.

The listed_in column can contain multiple genres for a single title.
I split these genre combinations into individual categories using split()
and explode(), then counted their frequency using value_counts().

This allows me to identify the most common genres in the Netflix dataset.
"""
```

```python
genres = df["listed_in"].str.split(", ").explode()
print(generes)
```

```python
genre_counts = genres.value_counts()
top_genres = genre_counts.head(10).reset_index()
print(top_genres)
```

```python
"""
Step 3: Compare ratings across content types.

I used a cross-tabulation to compare rating categories between Movies and
TV Shows. This allows me to examine how the distribution of content ratings
differs between the two content types.
"""
```

```python
rating_by_type = pd.crosstab(df["rating"], df["type"])
print(rating_by_type)
```

```python
# Step 4 — Interactive visualizations
```

```python
rating_df = rating_counts.reset_index()

rating_df.columns = ["rating", "content_count"]

fig = px.bar(
    rating_df,
    x="rating",
    y="content_count",
    title="Netflix Content Rating Distribution",
    labels={
        "rating": "Rating",
        "content_count": "Number of Titles"
    }
)

fig.show()
```

```python
# Top Genres
top_genres = genre_counts.head(10).reset_index()

top_genres.columns = ["genre", "content_count"]

fig = px.bar(
    top_genres,
    x="content_count",
    y="genre",
    orientation="h",
    title="Top 10 Netflix Genres",
    labels={
        "genre": "Genre",
        "content_count": "Number of Titles"
    }
)

fig.show()
```

```python
# Ratings by Content Type

rating_type_df = rating_by_type.reset_index()
print(rating_type_df)

fig = px.bar(
    rating_type_df,
    x="rating",
    y=["Movie", "TV Show"],
    barmode="group",
    title="Netflix Rating Distribution by Content Type",
    labels={
        "rating": "Rating",
        "value": "Number of Titles",
        "variable": "Content Type"
         }
)
fig.show()
```

```python
rating_percentage = (
    df["rating"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print(rating_percentage)
#TV-MA represents 36.46% of the content.
```

```python
genre_counts.head(10)
```

```python
genre_percentage = (
    genre_counts
    .div(len(df))
    .mul(100)
    .round(2)
)

print(genre_percentage.head(10))
```

```python
rating_by_type_percentage = pd.crosstab(
    df["rating"],
    df["type"],
    normalize="columns"
).mul(100).round(2)

print(rating_by_type_percentage)
#TV-MA represents  33.66% of TV Shows compared with  42.91% of Movies.
```
