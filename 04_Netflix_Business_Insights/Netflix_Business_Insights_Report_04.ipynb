```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
"""
#Exploratory Data Analysis(EDA)
Load → Preview → Shape → Data Types → Missing Values → Duplicates → Unique Values → Cleaning → Descriptive Statistics → Content Types → Ratings → Release Years → Countries → Genres → Date Analysis → Outliers/Unusual Values → Final Data Check
"""
```

```python
# Load the dataset & preview the data
df = pd.read_csv("/content/Dataset.csv")
print("THE TOP 10")
display(df.head(10))
print(" THE LAST 10")
display(df.tail(10))

```

```python
# Dataset shape
print("Number of rows :",df.shape[0])
print("Number of columns",df.shape[1])
print("Dataset dhape :",df.shape)
```

```python
# Column names and data types
print("Column Name")
print(df.columns.tolist())
print("\n Data Types")
print(df.dtypes)
```

```python
# More detailed
df.info()
```

```python
# Check missing values
missing_values = df.isnull().sum()
print("Missing values")
print(missing_values)

# presentage of missing values
print("\n")
missing_presentage = (missing_values/len(df))*100
print("Presentage of missig values")
print(missing_presentage)
```

```python
# Check duplicates records

print("Number of duplicate records:", df.duplicated().sum())

"""
thers is no duplicate records in the dataset
"""
```

```python
# Unique values

print(" Number of unique values in each column")
print(df.nunique())
```

```python
for column in ["type", "rating"]:
    if column in df.columns:
        print(f"\nUnique values in {column}:")
        print(df[column].unique())
```

```python
# Clean the Data
df.columns = df.columns.str.strip()
text_columns = df.select_dtypes(include = "object").columns
for column in text_columns:
  df[column] = df[ column].str.strip()
```

```python
if "date_added" in df.columns:
    df["date_added"] = pd.to_datetime(
        df["date_added"],
        errors="coerce"
    )
# Ceck the result
print(df.info())
```

```python
# Descriptive statistics
print("Descriptive statistics")

print("\n Descriptive numerical columns statistics")
display(df.describe())

print("\n Descriptive categorical columns statistics ")
display(df.describe(include = "object"))
print("\n Descriptive statistics for all dataset ")
print(df.describe(include = "all"))
```

```python
# Distribution Analysis
print("Distribution of content types")
if "type" in df.columns:
    type_counts = df["type"].value_counts()

    print(type_counts)

    plt.figure(figsize=(8, 5))
    sns.countplot(data=df, x="type")
    plt.title("Distribution of Content Types")
    plt.xlabel("Content Type")
    plt.ylabel("Count")
    plt.show()
```

```python
# Distribution of ratings
if "rating" in df.columns:
    rating_counts = df["rating"].value_counts()

    print(rating_counts)

    plt.figure(figsize=(10, 6))
    sns.countplot(
        data=df,
        y="rating",
        order=rating_counts.index
    )
    plt.title("Distribution of Ratings")
    plt.xlabel("Count")
    plt.ylabel("Rating")
    plt.show()
```

```python
# Distribution of release years
if "release_year" in df.columns:
    print(df["release_year"].describe())

    plt.figure(figsize=(12, 6))
    sns.histplot(
        data=df,
        x="release_year",
        bins=30
    )
    plt.title("Distribution of Release Years")
    plt.xlabel("Release Year")
    plt.ylabel("Number of Titles")
    plt.show()
```

```python
print("TOP 10 RELEASE YEARS")
print(df["release_year"].value_counts().head(10))
```

```python
# Distribution of countries
if "country" in df.columns:
    countries = (
        df["country"]
        .dropna()
        .str.split(",")
        .explode()
        .str.strip()
    )

    country_counts = countries.value_counts()

    print("TOP 10 COUNTRIES")
    print(country_counts.head(10))
    plt.figure(figsize=(12, 6))

sns.barplot(
    x=country_counts.head(10).values,
    y=country_counts.head(10).index
)

plt.title("Top 10 Countries by Number of Titles")
plt.xlabel("Number of Titles")
plt.ylabel("Country")
plt.show()
```

```python
# Distribution of genres/categories
if "listed_in" in df.columns:
    genres = (
        df["listed_in"]
        .dropna()
        .str.split(",")
        .explode()
        .str.strip()
    )

    genre_counts = genres.value_counts()

    print("TOP 10 GENRES/CATEGORIES")
    print(genre_counts.head(10))
    plt.figure(figsize=(12, 6))

sns.barplot(
    x=genre_counts.head(10).values,
    y=genre_counts.head(10).index
)

plt.title("Top 10 Genres/Categories")
plt.xlabel("Number of Titles")
plt.ylabel("Genre/Category")
plt.show()
```

```python
# Analyze date_added
if "date_added" in df.columns:

    print("Missing dates:", df["date_added"].isnull().sum())

    df["added_year"] = df["date_added"].dt.year
    df["added_month"] = df["date_added"].dt.month
    df["added_month_name"] = df["date_added"].dt.month_name()
```

```python
#Titles added per year
if "added_year" in df.columns:

    yearly_added = df["added_year"].value_counts().sort_index()

    print(yearly_added)

    plt.figure(figsize=(12, 6))
    yearly_added.plot(kind="bar")

    plt.title("Titles Added to Netflix by Year")
    plt.xlabel("Year")
    plt.ylabel("Number of Titles")
    plt.xticks(rotation=45)
    plt.show()
```

```python
#Titles added per month
if "added_month_name" in df.columns:

    month_order = [
        "January", "February", "March", "April",
        "May", "June", "July", "August",
        "September", "October", "November", "December"
    ]

    monthly_added = (
        df["added_month_name"]
        .value_counts()
        .reindex(month_order)
    )

    plt.figure(figsize=(12, 6))
    monthly_added.plot(kind="bar")

    plt.title("Titles Added by Month")
    plt.xlabel("Month")
    plt.ylabel("Number of Titles")
    plt.xticks(rotation=45)
    plt.show()
```

```python
# Check release-year anomalies
if "release_year" in df.columns:

    print("Minimum release year:", df["release_year"].min())
    print("Maximum release year:", df["release_year"].max())

    print("\nPotential unusual release years:")
    print(
        df[
            (df["release_year"] < 1900) |
            (df["release_year"] > 2026)
        ][["title", "release_year"]]
    )
```

```python
#Check numerical outliers
numeric_columns = df.select_dtypes(include="number").columns

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    print(f"\n{column}")
    print("Number of potential outliers:", len(outliers))
```

```python
#Final cleaned dataset check
print("FINAL DATASET SHAPE")
print(df.shape)

print("\nFINAL MISSING VALUES")
print(df.isnull().sum())

print("\nFINAL DUPLICATES")
print(df.duplicated().sum())

print("\nFINAL DATA TYPES")
print(df.dtypes)
```

```python
# Movies vs TV Shows over time
content_trend = (
    df.groupby(["release_year", "type"])
    .size()
    .reset_index(name="content")
)
plt.figure(figsize=(12,6))

sns.lineplot(
    data=content_trend,
    x="release_year",
    y="content",
    hue="type"
)
plt.title("Netflix Content Trends by Type and Release Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.show()
```

```python
# Top genres by content type
genre_type = (
    df[["type","listed_in"]]
    .dropna()
    .assign(listed_in=lambda x:x["listed_in"].str.split(","))
    .explode("listed_in")
)

genre_type["listed_in"] = genre_type["listed_in"].str.strip()
top_genres = (
    genre_type.groupby(["type","listed_in"])
    .size()
    .reset_index(name="count")
    .sort_values(by="count",ascending = False)
)
display(top_genres.head(20))
```

```python
# Country × Content Type Analysis

country_type = (
    df[["country", "type"]]
    .dropna()
    .assign(
        country=lambda x: x["country"].str.split(",")
    )
    .explode("country")
)

country_type["country"] = country_type["country"].str.strip()

country_type_counts = (
    country_type
    .groupby(["country", "type"])
    .size()
    .reset_index(name="count")
)

# Top 10 countries
top_countries = country_counts.head(10).index

display(country_type_counts.head(20))
```

```python
# Ratings by content type
rating_type = pd.crosstab(
    df["rating"],
    df["type"]
)

display(rating_type)
rating_type.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Ratings by Content Type")
plt.xlabel("Rating")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)
plt.show()
```

```python
# Content growth
content_growth =  (
    df.groupby("release_year")
    .size()
)
plt.figure(figsize=(12,6))
content_growth.plot()
plt.title("Netflix Content Growth Over Time")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.show()
```

```python
print("Year with the most content:")
print(content_growth.idxmax())

print("Maximum number of titles:")
print(content_growth.max())
```

```python
# Recent content analysis
recent_content = df[df["release_year"] >= 2015]

print("Recent content:")
print(recent_content.shape)

display(
    recent_content["type"]
    .value_counts()
)
```

```python
total_titles = len(df)
total_movies = (df["type"] == "Movie").sum()
total_tv = (df["type"] == "TV Show").sum()
total_countries = countries.nunique()
total_genres = genres.nunique()

print("Total Titles:", total_titles)
print("Movies:", total_movies)
print("TV Shows:", total_tv)
print("Countries:", total_countries)
print("Genres:", total_genres)
```

```python
fig, axes = plt.subplots(2, 2, figsize=(16, 10))

# Content type
df["type"].value_counts().plot(
    kind="bar",
    ax=axes[0, 0]
)

axes[0, 0].set_title("Movies vs TV Shows")
axes[0, 0].set_xlabel("Content Type")
axes[0, 0].set_ylabel("Titles")

# Top genres
genre_counts.head(10).sort_values().plot(
    kind="barh",
    ax=axes[0, 1]
)

axes[0, 1].set_title("Top 10 Genres")
axes[0, 1].set_xlabel("Titles")

# Top countries
country_counts.head(10).sort_values().plot(
    kind="barh",
    ax=axes[1, 0]
)

axes[1, 0].set_title("Top 10 Countries")
axes[1, 0].set_xlabel("Titles")

# Content growth
content_growth.plot(
    x="release_year",
    y="count",
    ax=axes[1, 1]
)

axes[1, 1].set_title("Content Growth")
axes[1, 1].set_xlabel("Release Year")
axes[1, 1].set_ylabel("Titles")

plt.suptitle(
    "Netflix Executive Overview Dashboard",
    fontsize=18
)

plt.tight_layout()
plt.show()
```

```python
# Global Content

fig, axes = plt.subplots(1, 2, figsize=(16, 7))

# Top 10 countries
country_counts.head(10).sort_values().plot(
    kind="barh",
    ax=axes[0]
)

axes[0].set_title("Top 10 Content-Producing Countries")
axes[0].set_xlabel("Number of Titles")
axes[0].set_ylabel("Country")

# Country × Content Type
country_pivot = (
    country_type_counts[
        country_type_counts["country"].isin(top_countries)
    ]
    .pivot(
        index="country",
        columns="type",
        values="count"
    )
    .fillna(0)
)

country_pivot.plot(
    kind="bar",
    ax=axes[1]
)

axes[1].set_title("Country Content by Type")
axes[1].set_xlabel("Country")
axes[1].set_ylabel("Number of Titles")

plt.suptitle(
    "Netflix Global Content Dashboard",
    fontsize=18
)

plt.tight_layout()
plt.show()
```
